```python
import pandas as pd

df = pd.read_csv(r"C:\Users\TRISTAR\Downloads\pharmaly_products_800.csv")
```

```python
# Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# Check duplicate product URLs
print("Duplicate product URLs:", df["product_url"].duplicated().sum())

# Check empty strings
for column in df.columns:
    print(column, ":", (df[column].str.strip() == "").sum())
```

```python
# Display some product names
print(df["product_name"].head(15).to_list())

# Display brands
print(df["brand"].value_counts().head(15))

# Display categories
print(df["categories"].value_counts().head(15))

# Display some prices
print(df[["original_price", "current_price"]].head(15))
```

```python
text_columns = [
    "product_name",
    "brand",
    "categories",
    "short_description",
    "full_description"
]

for column in text_columns:
    df[column] = df[column].str.replace(r"\s+", " ", regex=True).str.strip()
```

```python
df["product_name"].head(15)
```

```python
print("Current prices:")
print(df["current_price"].value_counts().head(20))

print("\nOriginal prices:")
print(df["original_price"].dropna().value_counts().head(20))
```

```python
def clean_price(price):
    if pd.isna(price):
        return None

    price = price.replace("DA", "").strip()
    price = price.replace(".", "")
    price = price.replace(",", ".")

    return float(price)


df["current_price"] = df["current_price"].apply(clean_price)
df["original_price"] = df["original_price"].apply(clean_price)
```
